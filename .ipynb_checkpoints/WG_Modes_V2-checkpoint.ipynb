#### Note that the transverse mode axis are Y-Z, with Z being the height.

#### This means the surface normal to the mode is in the x direction! e.g. $$ d\vec{a} = d\vec{x} $$

# Initialization code

```python
def linPlotMode(Mi,num,yes):
    if yes:
        Md = Mi[num-1][0]
        print(Mi[num-1])
        Ey = Md.E.y
        x = Md.E.yy
        y = Md.E.zz
        xx, yy = np.meshgrid(x,y) 
        plt.rcParams["figure.figsize"] = (8,5) 
        plt.subplot() 
        plt.pcolormesh(xx,yy, np.transpose(np.absolute(Ey)**2), cmap='jet') 
        plt.axis([x.min(), x.max(), y.min(), y.max()]) 
        plt.colorbar()  
        plt.show()
        
def logPlotMode(Mi,num,yes):
    if yes:
        Md = Mi[num-1][0]
        Ey = Md.E.y
        
        x = Md.E.yy
        y = Md.E.zz
        xx, yy = np.meshgrid(x,y) 
        plt.rcParams["figure.figsize"] = (8,5) 
        plt.subplot() 
        plt.pcolormesh(xx,yy, np.transpose(10*np.log10(np.absolute(Ey)**2)), cmap='jet') 
        plt.axis([x.min(), x.max(), y.min(), y.max()]) 
        plt.colorbar()  
        plt.show()
        
def linPlotIndex(ind,x,y):
        index = ind
        xx, yy = np.meshgrid(x,y) 
        plt.rcParams["figure.figsize"] = (8,5) 
        plt.subplot() 
        plt.pcolormesh(xx,yy, np.transpose(index), cmap='jet') 
#         plt.axis([x.min(), x.max(), y.min(), y.max()]) 
        plt.colorbar()  
        plt.show()
        
# Overlap compares the overlap of a gaussian profile from python and a Lumerical profile
def overlapLUM(ExWG1,ExWG2,din,x,y):
    Ext1 = np.transpose(ExWG1)
    Ext2 = np.transpose(ExWG2)
    din = np.transpose(din)
    

    numIntegrand = np.conjugate(Ext1)*Ext2*din
    denIntegrandWG = np.absolute(Ext1)**2
    denIntegrandMODE = np.absolute(Ext2)**2

    numInt = (simps(simps(numIntegrand, x), y))
    denIntMODE = simps(simps(denIntegrandMODE, x), y)
    denIntWG = simps(simps(denIntegrandWG, x), y)

    overlap = numInt / np.sqrt(denIntMODE*denIntWG)
    return overlap

# OverlapLUM compares the overlap of two Lumerical profiles
def overlap(ea,eb,ϵa,ϵ,ω,x,y):
    num = np.conjugate(ea)*eb
    numInt = simps(simps(num,y),x)
    
    den = np.absolute(ea)**2
    denInt = simps(simps(den,y),x)
    
    integral = numInt / denInt
    return integral

def κab(ea,eb,ϵa,ϵ,ω,x,y):
    ϵ0 = 8.85418782E-12
    num = (ω/4)*ϵ0*ϵa*np.conjugate(ea)*eb
    numInt = simps(simps(num,y),x)
    
    den = (ϵ0/4)*ϵ*np.absolute(ea)**2
    denInt = simps(simps(den,y),x)
    
    integral = numInt / denInt
    
    return integral
```

# BENDING LOSS

```python
#Parameters for simulation

mesh = 100*nm
l_wg = 0.1*um
chip_width = 60*um

wl = 1550*nm
ω = 2*pi*(c/wl)

h_LN = 700*nm
h_etch = 350*nm
h_substrate = 4.7*um
h_cladding = 800*nm
w_wg = 1000*nm

angle_sw = 60*deg
trim = h_etch/np.tan(angle_sw)
offset = 0

'''Set bend to true and enter a radius to perform mode simulations for a bent waveguide'''
bend = True
rd = 1000*um
```

```python
''' Check to make sure settings in lumerical are correctly set by following code '''

mds.defineGlobals(wl, h_LN, h_etch, h_substrate, h_cladding, w_wg, angle_sw, chip_width)
mds.updateMaterial('y', 21.5)
mds.drawWG(l_wg, offset)
mds.setupFDE(width_fde = 53e-6, size_fde_mesh = 100e-9, width_med = 5e-6, size_med_mesh = 10e-9, width_fine = 2.5e-6, size_fine_mesh = 2e-9)
# M_test, in_test = mds.solve_bent_modes(rd,10, bent = bend)
Ma, ina = mds.solve_bent_modes_near(rd, n1, n2, nmodes=10, bent = False)
```

```python
radii = np.linspace(300*um,50*um,21)
print(radii)
```

```python
''' Test the minimum bend radius to ensure a TE mode still exists'''
rr = radii[0]

mds.defineGlobals(wl, h_LN, h_etch, h_substrate, h_cladding, w_wg, angle_sw, chip_width)
mds.updateMaterial('y', 21.5)
mds.drawWG(l_wg, offset)
mds.setupFDE(width_fde = 53e-6, size_fde_mesh = 100e-9, width_med = 5e-6, size_med_mesh = 10e-9, width_fine = 2.5e-6, size_fine_mesh = 2e-9)

## Modes are listed in order of descending TEFP
# M_test, in_test = mds.solve_bent_modes(rr,10, bent = bend)
Ma, ina = mds.solve_bent_modes_at_n(rr, 1.95, nmodes=10, bent = bend)

lum_modes = np.array(Ma)
threshold = 0.6
TE_modes = lum_modes[(lum_modes[:,1] > threshold)]
print(np.shape(TE_modes))
if np.shape(TE_modes)[0] == 0:
    print("No modes found at "+str(rr)+" µm bend radius")
else:
    fundamental_TE = TE_modes[0]
    fundamental_TE_idx = np.where(lum_modes[:,1] == fundamental_TE[:][1])[0][0]
    lss = Ma[fundamental_TE_idx][3]
    print("Fundamental mode loss: "+str(round(lss,9))+" [dB/m]")
```

```python
fps = [0]*len(radii)
nfs = [0]*len(radii)
lss = [0]*len(radii)
ti = time.time()

for i, rr in enumerate(radii):
    mds.defineGlobals(wl, h_LN, h_etch, h_substrate, h_cladding, w_wg, angle_sw, chip_width)
    mds.updateMaterial('y', 21.5)
    mds.drawWG(l_wg, offset)
    mds.setupFDE(width_fde = 53e-6, size_fde_mesh = 100e-9, width_med = 5e-6, size_med_mesh = 10e-9, width_fine = 2.5e-6, size_fine_mesh = 2e-9)
    
    # Modes are listed in order of descending TEFP
#     Ma, ina = mds.solve_bent_modes(rr,10, bent = bend)
    Ma, ina = mds.solve_bent_modes_at_n(rr, 1.95, nmodes=10, bent = bend)

    lum_modes = np.array(Ma)
    threshold = 0.6
    TE_modes = lum_modes[(lum_modes[:,1] > threshold)]
    if np.shape(TE_modes)[0] == 0:
        print("No TE modes found at "+str(rr)+" µm bend radius.")
    else:
        fundamental_TE = TE_modes[0]
        fundamental_TE_idx = np.where(lum_modes[:,1] == fundamental_TE[:][1])[0][0]
        lss[i] = Ma[fundamental_TE_idx][3]
        if lss[i] < 1e-6:
            lss[i] = 0
        fps[i] = Ma[fundamental_TE_idx][1]
        print("Fundamental mode loss: "+str(round(lss[i],9))+" [dB/m]")
        
    
tf = time.time()-ti
print(tf/60)
```

```python
radii_new = np.flip(np.linspace(300*um,50*um,50))
fit = CubicSpline(np.flip(radii), np.flip(lss), axis = 0, bc_type='natural')
lss_new = fit(radii_new)
```
