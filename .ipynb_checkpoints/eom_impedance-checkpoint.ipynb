## Assumptions
- w_g >= 3 w_s where w_g is width of ground electrode, w_s width of signal electrode
- h > s (thickness of substrate > gap) 
- all units in um

```python
import numpy as np

# h = substrate thickness, s = gap, k = w/w+2s where w = width of ground electrode
# frequency independent?
def effective_permittivity(epsilon_r, h, s, k):
    term1 = 0.5 * (epsilon_r + 1)
    log_term = np.log10(h / s)
    tanh_term = np.tanh(1.785 * log_term + 1.75)
    addition_term = (k * s / h) * (0.04 - 0.7 * k + 0.01 * (1 - 0.1 * epsilon_r) * (0.25 + k))
    epsilon_eff = term1 * (tanh_term + addition_term)
    return epsilon_eff

# used in effective permittivity calculation, not sure what this physically represents
def k(w_s, s):
    return w_s / (w_s + 2 * s)

# approximation of effective permittivity (average of air and substrate permittivity)
def effective_permittivity_approx(epsilon_r):
    return (epsilon_r + 1) / 2

# impedance calculation given effective permittivity and k
def impedance(epsilon_eff, k):
    sqrt_k = np.sqrt(k)
    elliptic_integral_ratio = (1 / np.pi) * np.log(2 * (1 + sqrt_k) / (1 - sqrt_k))
    z0 = (30 * np.pi / np.sqrt(epsilon_eff)) * elliptic_integral_ratio
    return z0

def relative_permittivity():
    return
# NOTE: analysis is a good approximation if w_g >= 3 * w_s
w_s = 8 # width of signal electrode
w_g = 28 # width of ground electrode (not actually used for computations)

epsilon_r = 3.9 #SiO2 relative permittivity
h = 5 # thickness of substrate
s = 4 # gap between electrodes
k = k(w_s, s)
epsilon_e = effective_permittivity(epsilon_r, h, s, k)
z0 = impedance(epsilon_e, k)

# compare approx vs analytic solution
# print("Effective permittivity approximation:", effective_permittivity_approx(epsilon_r))
print("Effective permittivity computation:", epsilon_e)
print("Impedance:", z0)

# XXTODO: Given impedance of 50 Ohm, back calculate for dimensions?
```
